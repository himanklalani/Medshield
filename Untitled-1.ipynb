```python
import os

print("=" * 80)
print("CREATING PYTHON MODULES")
print("=" * 80)

# Create working directories if needed
os.makedirs(r'D:\himank model\models', exist_ok=True)
os.makedirs(r'D:\himank model\federated_data', exist_ok=True)

# ===== MODULE 1: DATA PREPARATION =====
data_prep_code = r'''
import os
import numpy as np
import shutil
from sklearn.model_selection import train_test_split
from glob import glob
import random

def partition_data_for_federated_learning(
    data_dir=r'D:\himank model\brain_tumor_data',
    output_dir=r'D:\himank model\federated_data',
    num_clients=2,
    random_seed=42
):
    random.seed(random_seed)
    np.random.seed(random_seed)

    classes = ['Brain_Glioma', 'Brain_Menin', 'Brain_Tumor']

    print(f"\n{'='*60}")
    print(f"PARTITIONING DATA FOR {num_clients} HOSPITALS")
    print(f"{'='*60}")

    # Create client directories
    for client_id in range(num_clients):
        client_dir = os.path.join(output_dir, f'hospital_{client_id + 1}')
        for split in ['train', 'val', 'test']:
            for class_name in classes:
                split_dir = os.path.join(client_dir, split, class_name)
                os.makedirs(split_dir, exist_ok=True)

    # Partition each class
    for class_name in classes:
        class_path = os.path.join(data_dir, class_name)
        image_files = []
        for ext in ['*.jpg', '*.jpeg', '*.png', '*.bmp']:
            image_files.extend(glob(os.path.join(class_path, ext)))
        random.shuffle(image_files)
        print(f"\nClass: {class_name}")
        print(f"  Total images: {len(image_files)}")
        images_per_client = len(image_files) // num_clients

        for client_id in range(num_clients):
            start_idx = client_id * images_per_client
            end_idx = start_idx + images_per_client if client_id < num_clients - 1 else len(image_files)
            client_images = image_files[start_idx:end_idx]
            train_imgs, temp_imgs = train_test_split(client_images, test_size=0.3, random_state=random_seed)
            val_imgs, test_imgs = train_test_split(temp_imgs, test_size=0.5, random_state=random_seed)
            client_dir = os.path.join(output_dir, f'hospital_{client_id + 1}')
            for img_path in train_imgs:
                dst = os.path.join(client_dir, 'train', class_name, os.path.basename(img_path))
                shutil.copy2(img_path, dst)
            for img_path in val_imgs:
                dst = os.path.join(client_dir, 'val', class_name, os.path.basename(img_path))
                shutil.copy2(img_path, dst)
            for img_path in test_imgs:
                dst = os.path.join(client_dir, 'test', class_name, os.path.basename(img_path))
                shutil.copy2(img_path, dst)
            print(f"  Hospital {client_id + 1}: Train={len(train_imgs)}, Val={len(val_imgs)}, Test={len(test_imgs)}")

    print(f"\n{'='*60}")
    print("DATA PARTITIONING COMPLETED!")
    print(f"{'='*60}")

if __name__ == "__main__":
    partition_data_for_federated_learning()
'''

with open(r'D:\himank model\data_preparation.py', 'w', encoding='utf-8') as f:
    f.write(data_prep_code)
print("✓ data_preparation.py created")


# ===== MODULE 2: MODEL ARCHITECTURE =====
model_code = r'''
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2

def create_cnn_model(input_shape=(224, 224, 3), num_classes=3):
    base_model = MobileNetV2(
        input_shape=input_shape,
        include_top=False,
        weights='imagenet'
    )
    base_model.trainable = False
    model = models.Sequential([
        layers.Input(shape=input_shape),
        layers.Rescaling(1./255),
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.BatchNormalization(),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.5),
        layers.BatchNormalization(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(num_classes, activation='softmax')
    ])
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
                  loss='categorical_crossentropy', metrics=['accuracy'])
    return model
'''

with open(r'D:\himank model\model_architecture.py', 'w', encoding='utf-8') as f:
    f.write(model_code)
print("✓ model_architecture.py created")


# ===== MODULE 3: FEDERATED SERVER =====
server_code = r'''
import numpy as np
import tensorflow as tf
from typing import List

class FederatedServer:
    def __init__(self, model):
        self.global_model = model
        self.round_history = []

    def get_global_weights(self):
        return self.global_model.get_weights()

    def federated_averaging(self, client_weights_list: List, client_data_sizes: List):
        total_samples = sum(client_data_sizes)
        num_layers = len(client_weights_list[0])
        aggregated_weights = []
        for layer_idx in range(num_layers):
            layer_weights = np.zeros_like(client_weights_list[0][layer_idx], dtype=np.float64)
            for client_idx, client_weights in enumerate(client_weights_list):
                weight = client_data_sizes[client_idx] / total_samples
                layer_weights += weight * client_weights[layer_idx]
            aggregated_weights.append(layer_weights)
        return aggregated_weights

    def aggregate_and_update(self, client_weights_list: List, client_data_sizes: List, round_num: int):
        print(f"\n{'='*60}")
        print(f"Round {round_num}: Aggregating {len(client_weights_list)} hospital updates")
        print(f"{'='*60}")
        aggregated_weights = self.federated_averaging(client_weights_list, client_data_sizes)
        self.global_model.set_weights(aggregated_weights)
        print("Global model updated")

    def evaluate_global_model(self, test_data, round_num: int):
        results = self.global_model.evaluate(test_data, verbose=0)
        metrics = {'round': round_num, 'loss': results[0], 'accuracy': results[1]}
        self.round_history.append(metrics)
        print(f"Global Model Performance:")
        print(f"  Loss: {results[0]:.4f}")
        print(f"  Accuracy: {results[1]:.4f} ({results[1]*100:.2f}%)")
        return metrics
'''

with open(r'D:\himank model\federated_server.py', 'w', encoding='utf-8') as f:
    f.write(server_code)
print("✓ federated_server.py created")


# ===== MODULE 4: FEDERATED CLIENT =====
client_code = r'''
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

class FederatedClient:
    def __init__(self, client_id: int, data_dir: str, model, batch_size: int = 32, img_size: tuple = (224, 224)):
        self.client_id = client_id
        self.data_dir = data_dir
        self.local_model = model
        self.batch_size = batch_size
        self.img_size = img_size
        self.train_data, self.train_samples = self._load_data('train')
        self.val_data, self.val_samples = self._load_data('val')
        self.test_data, self.test_samples = self._load_data('test')
        print(f"Hospital {client_id} Initialized:")
        print(f"   Train: {self.train_samples:,} images")
        print(f"   Val: {self.val_samples:,} images")
        print(f"   Test: {self.test_samples:,} images")

    def _load_data(self, split: str):
        if split == 'train':
            datagen = ImageDataGenerator(
                rotation_range=20, width_shift_range=0.2, height_shift_range=0.2,
                horizontal_flip=True, zoom_range=0.2, shear_range=0.15, fill_mode='nearest'
            )
        else:
            datagen = ImageDataGenerator()
        data_path = os.path.join(self.data_dir, split)
        data = datagen.flow_from_directory(
            data_path, target_size=self.img_size, batch_size=self.batch_size,
            class_mode='categorical', shuffle=(split == 'train')
        )
        return data, data.n

    def update_local_model(self, global_weights):
        self.local_model.set_weights(global_weights)

    def train_local_model(self, epochs: int = 5, verbose: int = 1):
        print(f"{'─'*60}")
        print(f"Hospital {self.client_id}: Local Training")
        print(f"{'─'*60}")
        history = self.local_model.fit(
            self.train_data, epochs=epochs, validation_data=self.val_data, verbose=verbose,
            steps_per_epoch=max(1, self.train_samples // self.batch_size),
            validation_steps=max(1, self.val_samples // self.batch_size)
        )
        updated_weights = self.local_model.get_weights()
        print(f"Hospital {self.client_id} training complete")
        return history, self.train_samples, updated_weights

    def evaluate_local_model(self):
        results = self.local_model.evaluate(
            self.test_data, steps=max(1, self.test_samples // self.batch_size), verbose=0
        )
        print(f"   Hospital {self.client_id} Test - Loss: {results[0]:.4f}, Accuracy: {results[1]:.4f}")
        return results
'''

with open(r'D:\himank model\federated_client.py', 'w', encoding='utf-8') as f:
    f.write(client_code)
print("✓ federated_client.py created")

print("\n" + "=" * 80)
print("✅ ALL MODULES CREATED SUCCESSFULLY!")
print("=" * 80)
print("\nFiles created:")
print("  • data_preparation.py")
print("  • model_architecture.py")
print("  • federated_server.py")
print("  • federated_client.py")
```

```python
# ============================================================================
# CELL 5: TRAIN WITH LOCAL WINDOWS TENSORFLOW
# ============================================================================

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import warnings
import os
import time
import json
import sys

warnings.filterwarnings('ignore')
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

print(f"Using TensorFlow: {tf.__version__}")
print(f"Using NumPy: {np.__version__}")

# Add your modules folder if needed
sys.path.append(r'D:\himank model')

# Import your modules
from model_architecture import create_cnn_model
from federated_server import FederatedServer
from federated_client import FederatedClient

print("\n" + "="*80)
print("FEDERATED LEARNING TRAINING")
print("="*80)

# Configuration
NUM_ROUNDS = 10
LOCAL_EPOCHS = 5
BATCH_SIZE = 32
NUM_HOSPITALS = 2

print(f"\nConfiguration:")
print(f"  Federated Rounds: {NUM_ROUNDS}")
print(f"  Local Epochs per Round: {LOCAL_EPOCHS}")
print(f"  Batch Size: {BATCH_SIZE}")
print(f"  Number of Hospitals: {NUM_HOSPITALS}")
print("="*80)

start_time = time.time()

# [1/5] Create model
print("\n[1/5] Creating global CNN model...")
global_model = create_cnn_model()
print(f"✓ Model created - Parameters: {global_model.count_params():,}")

# [2/5] Initialize server
print("\n[2/5] Initializing Federated Server...")
server = FederatedServer(global_model)
print("✓ Server initialized")

# [3/5] Initialize hospitals
print("\n[3/5] Initializing Hospital Nodes...")
clients = []

for i in range(NUM_HOSPITALS):
    print(f"\nInitializing Hospital {i+1}...")

    client_model = tf.keras.models.clone_model(global_model)
    client_model.set_weights(global_model.get_weights())
    client_model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    client = FederatedClient(
        client_id=i + 1,
        data_dir=fr'D:\himank model\federated_data\hospital_{i + 1}',
        model=client_model,
        batch_size=BATCH_SIZE
    )
    clients.append(client)

print("\n✓ All hospitals initialized")

# [4/5] Training loop
print("\n[4/5] Starting Federated Training...")
print("="*80)
print("This will take 2-3 hours. Keep your system active!")
print("="*80)

history = {
    'rounds': [],
    'global_accuracy': [],
    'global_loss': [],
    'hospital_1_acc': [],
    'hospital_2_acc': [],
    'time_per_round': []
}

for round_num in range(1, NUM_ROUNDS + 1):
    round_start = time.time()

    print(f"\n{'#'*80}")
    print(f"# FEDERATED ROUND {round_num}/{NUM_ROUNDS}")
    print(f"{'#'*80}")

    global_weights = server.get_global_weights()

    client_weights_list = []
    client_data_sizes = []

    for client in clients:
        client.update_local_model(global_weights)

        hist, num_samples, updated_weights = client.train_local_model(
            epochs=LOCAL_EPOCHS,
            verbose=1
        )

        client_weights_list.append(updated_weights)
        client_data_sizes.append(num_samples)

        client_results = client.evaluate_local_model()
        history[f'hospital_{client.client_id}_acc'].append(client_results[1])

    server.aggregate_and_update(
        client_weights_list,
        client_data_sizes,
        round_num
    )

    global_metrics = server.evaluate_global_model(
        clients[0].test_data,
        round_num
    )

    history['rounds'].append(round_num)
    history['global_accuracy'].append(global_metrics['accuracy'])
    history['global_loss'].append(global_metrics['loss'])

    round_time = time.time() - round_start
    history['time_per_round'].append(round_time)

    print(f"\n{'='*80}")
    print(f"ROUND {round_num} SUMMARY")
    print(f"{'='*80}")
    print(f"  Global Accuracy: {global_metrics['accuracy']*100:.2f}%")
    print(f"  Global Loss: {global_metrics['loss']:.4f}")
    print(f"  Hospital 1: {history['hospital_1_acc'][-1]*100:.2f}%")
    print(f"  Hospital 2: {history['hospital_2_acc'][-1]*100:.2f}%")
    print(f"  Round Time: {round_time/60:.2f} minutes")
    print(f"  Estimated Remaining: {(NUM_ROUNDS-round_num)*round_time/60:.1f} minutes")
    print(f"{'='*80}")

# [5/5] Save results
print("\n[5/5] Saving Results...")

os.makedirs(r'D:\himank model\models', exist_ok=True)

model_path = r'D:\himank model\models\federated_brain_tumor_model.h5'
server.global_model.save(model_path)
print(f"✓ Model saved: {model_path}")

with open(r'D:\himank model\models\training_history.json', 'w', encoding='utf-8') as f:
    json.dump(history, f, indent=4)
print("✓ Training history saved")

total_time = time.time() - start_time

print("\n" + "="*80)
print("FEDERATED LEARNING TRAINING COMPLETED!")
print("="*80)
print(f"Total Training Time: {total_time/3600:.2f} hours ({total_time/60:.1f} minutes)")
print(f"Final Global Accuracy: {history['global_accuracy'][-1]*100:.2f}%")
print(f"Final Global Loss: {history['global_loss'][-1]:.4f}")
print("="*80)

# Plot results
print("\nGenerating training plots...")

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

axes[0].plot(history['rounds'], [a*100 for a in history['global_accuracy']],
             'b-o', linewidth=2, markersize=8, label='Global Model')
axes[0].plot(history['rounds'], [a*100 for a in history['hospital_1_acc']],
             'r--s', linewidth=1.5, markersize=6, label='Hospital 1')
axes[0].plot(history['rounds'], [a*100 for a in history['hospital_2_acc']],
             'g--s', linewidth=1.5, markersize=6, label='Hospital 2')
axes[0].set_xlabel('Federated Round', fontsize=12)
axes[0].set_ylabel('Accuracy (%)', fontsize=12)
axes[0].set_title('Model Accuracy Over Federated Rounds', fontsize=14, fontweight='bold')
axes[0].legend(fontsize=10)
axes[0].grid(True, alpha=0.3)
axes[0].set_ylim([0, 100])

axes[1].plot(history['rounds'], history['global_loss'],
             'b-o', linewidth=2, markersize=8)
axes[1].set_xlabel('Federated Round', fontsize=12)
axes[1].set_ylabel('Loss', fontsize=12)
axes[1].set_title('Global Model Loss Over Federated Rounds', fontsize=14, fontweight='bold')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(r'D:\himank model\models\training_progress.png', dpi=300, bbox_inches='tight')
print("✓ Plot saved: D:\\himank model\\models\\training_progress.png")
plt.show()

print("\nAll files saved successfully!")
print("\nGenerated files:")
print("  • D:\\himank model\\models\\federated_brain_tumor_model.h5")
print("  • D:\\himank model\\models\\training_history.json")
print("  • D:\\himank model\\models\\training_progress.png")
print("\nNext: Run Cell 6 to create the Streamlit UI!")
```

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import numpy as np
import tensorflow as tf

# Load model
model = tf.keras.models.load_model("models/federated_brain_tumor_model.h5")

# Load image and preprocess
img_path = r"D:\himank model\Brain_Tumor\brain_tumor_0023.jpg"
img = load_img(img_path, target_size=(224, 224))  # no augmentations here
arr = img_to_array(img)  # (224, 224, 3)
arr = np.expand_dims(arr, axis=0).astype('float32')  # add batch dimension

# Do not rescale here; model applies rescaling internally
pred = model.predict(arr)
predicted_class_idx = np.argmax(pred[0])

print(f"Predicted class index: {predicted_class_idx}")
print(f"Confidence: {pred[0][predicted_class_idx] * 100:.2f}%")
```
